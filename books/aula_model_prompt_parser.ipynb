## Chat API : OpenAI

Let's start with a direct API calls to OpenAI.

```python

client = openai.OpenAI()

def get_completion(prompt, model=llm_model):
    messages = [
        {
        "role": "user",
        "content": prompt
        }
    ]

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0,
    )

    return response.choices[0].message.content
```

```python
client = openai.OpenAI()

with client.chat.completions.stream(
    model=llm_model,
    messages=[{"role": "user", "content": "Explique brevemente o que é aprendizado de máquina."}],
    temperature=0,
) as stream:
    for event in stream:
        if event.type == "message.delta":
            print(event.delta, end="", flush=True)
```

### Format output

```python
from pydantic import BaseModel, Field

class EmailResponse(BaseModel):
    customer_email: str = Field(..., description="The email content generated for the customer.")
    tips: str = Field(..., description="Additional tips for improving the email.")
    chain_of_thought: str = Field(..., description="The reasoning process behind the email generation.")
    score: int = Field(..., description="A score from 1 to 10 evaluating the quality of the email.")
```

```python
email_writer_chain = (
    chat
    .with_structured_output(EmailResponse)
)

prompt.format(
    customer_email=customer_email,
    style=styles[0],
    tone=tones[0],
    date=get_today_str()
)

response = email_writer_chain.invoke([HumanMessage(content=prompt)])
print("DONE")
```

### Exercicio

Para aprimorar os resultados obtidos vamos a criar um sistemas multi-agente para conseguir fazer uma análise do problema (reflexão), escrever o email a partir deste plano (escrita) e finalmente avaliar os resultados obtidos (avaliação). 

- Step 1: Crie os três agentes (Agente reflexão, Agente escrita, Agente avaliação)
- Step 2: Crie um pipeline (chain) para gerar os resultados
- Step 3: Avalie e intere sobre os resultados obtidos


```python
from pydantic import BaseModel, Field
from typing import List, Optional

# Modelo para o output do agente de reflexão
class ReflectionOutput(BaseModel):
    analysis: str = Field(..., description="Análise detalhada do problema e contexto")
    key_points: List[str] = Field(..., description="Pontos chave a serem abordados no email")
    strategy: str = Field(..., description="Estratégia sugerida para escrita do email")
    tone_suggestions: List[str] = Field(..., description="Sugestões específicas de tom e estilo")

# Modelo para o output do agente de escrita
class WritingOutput(BaseModel):
    email_draft: str = Field(..., description="Rascunho do email gerado")
    reasoning: str = Field(..., description="Raciocínio por trás das escolhas de escrita")
    followed_points: List[str] = Field(..., description="Pontos-chave que foram seguidos")

# Modelo para o output do agente de avaliação
class EvaluationOutput(BaseModel):
    score: int = Field(..., description="Pontuação de 1-10 para o email")
    strengths: List[str] = Field(..., description="Pontos fortes do email")
    weaknesses: List[str] = Field(..., description="Pontos a melhorar")
    suggestions: List[str] = Field(..., description="Sugestões de melhoria")
    approval: bool = Field(..., description="Se o email está aprovado ou precisa revisão")

# Criando os agentes especializados
reflection_agent = (
    chat.with_structured_output(ReflectionOutput)
)

writing_agent = (
    chat.with_structured_output(WritingOutput)
)

evaluation_agent = (
    chat.with_structured_output(EvaluationOutput)
)
```

```python
# Definindo os prompts para cada agente
reflection_prompt = """
Você é um agente especializado em análise estratégica de comunicação corporativa.
Sua tarefa é analisar o contexto fornecido e desenvolver uma estratégia para a escrita de um email.

<Context>
{customer_email}
Estilo: {style}
Tom: {tone}
Data: {date}
</Context>

Por favor, forneça:
1. Uma análise detalhada do problema e contexto
2. Os pontos-chave que devem ser abordados
3. Uma estratégia sugerida para a escrita
4. Sugestões específicas de tom e estilo
"""

writing_prompt = """
Você é um agente especializado em redação corporativa.
Use a análise e estratégia fornecidas para criar um email efetivo.

<Analysis>
{reflection_output}
</Analysis>

<Context>
{customer_email}
Estilo: {style}
Tom: {tone}
Data: {date}
</Context>

Por favor, escreva o email seguindo a estrutura:
1. Saudação
2. Introdução ao problema
3. Apresentação da solução
4. Benefícios da solução
5. Chamada para ação (CTA)
6. Despedida
"""

evaluation_prompt = """
Você é um agente especializado em avaliação de comunicação corporativa.
Sua tarefa é avaliar o email gerado e fornecer feedback construtivo.

<Original Context>
{customer_email}
Estilo: {style}
Tom: {tone}
</Original Context>

<Generated Email>
{email_draft}
</Generated Email>

Por favor, avalie:
1. Adequação ao estilo e tom solicitados (1-10)
2. Pontos fortes do email
3. Pontos que precisam de melhoria
4. Sugestões específicas para aprimoramento
5. Decisão final: o email está pronto para envio ou precisa de revisão?
"""

# Função para executar o pipeline completo
def run_email_pipeline(customer_email: str, style: str, tone: str, date: str, iterations: int = 1) -> dict:
    results = []

    for i in range(iterations):
        print(f"\nIteração {i+1}:")

        # Step 1: Reflexão
        reflection_message = reflection_prompt.format(
            customer_email=customer_email,
            style=style,
            tone=tone,
            date=date
        )
        reflection_result = reflection_agent.invoke(
            [HumanMessage(content=reflection_message)]
        )
        print("✓ Análise concluída")

        # Step 2: Escrita
        writing_message = writing_prompt.format(
            reflection_output=reflection_result.model_dump_json(),
            customer_email=customer_email,
            style=style,
            tone=tone,
            date=date
        )
        writing_result = writing_agent.invoke(
            [HumanMessage(content=writing_message)]
        )
        print("✓ Email gerado")

        # Step 3: Avaliação
        evaluation_message = evaluation_prompt.format(
            customer_email=customer_email,
            style=style,
            tone=tone,
            email_draft=writing_result.email_draft
        )
        evaluation_result = evaluation_agent.invoke(
            [HumanMessage(content=evaluation_message)]
        )
        print("✓ Avaliação concluída")

        results.append({
            'iteration': i+1,
            'reflection': reflection_result,
            'writing': writing_result,
            'evaluation': evaluation_result
        })

        # Se o email foi aprovado, podemos parar as iterações
        if evaluation_result.approval:
            print(f"\nEmail aprovado na iteração {i+1}!")
            break

    return results
```

```python
# Testando o pipeline com o exemplo anterior
results = run_email_pipeline(
    customer_email=customer_email,
    style=styles[0],  # formal and technical
    tone=tones[0],    # confident
    date=get_today_str(),
    iterations=2      # Permitir até 2 iterações caso necessário
)

# Analisando os resultados
for result in results:
    print(f"\n=== Iteração {result['iteration']} ===")

    print("\nReflexão:")
    print("Análise:", result['reflection'].analysis)
    print("\nPontos-chave:")
    for point in result['reflection'].key_points:
        print(f"- {point}")

    print("\nEmail Gerado:")
    print(result['writing'].email_draft)

    print("\nAvaliação:")
    print(f"Pontuação: {result['evaluation'].score}/10")
    print("\nPontos fortes:")
    for strength in result['evaluation'].strengths:
        print(f"- {strength}")
    print("\nSugestões de melhoria:")
    for suggestion in result['evaluation'].suggestions:
        print(f"- {suggestion}")

    print(f"\nAprovado: {'Sim' if result['evaluation'].approval else 'Não'}")
```
